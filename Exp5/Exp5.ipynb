```python
import os
import sys
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from matplotlib import font_manager

parent_dir = os.path.abspath(os.path.join(os.getcwd(), ".."))
if parent_dir not in sys.path:
    sys.path.append(parent_dir)

from Exp1.Exp1 import perform_eda
```

```python
import os

# Folder to save all plots as high-resolution EPS files
IMAGES_DIR = "images"
os.makedirs(IMAGES_DIR, exist_ok=True)

def save_fig(filename, fig=None):
    """Save the current (or given) figure as a 600 dpi EPS file inside the images folder."""
    path = os.path.join(IMAGES_DIR, filename)
    if fig is not None:
        fig.savefig(path, format='eps', dpi=600, bbox_inches='tight')
    else:
        plt.savefig(path, format='eps', dpi=600, bbox_inches='tight')
    print(f"Saved: {path}")
```

```python
# Perform EDA
df = pd.read_csv("wdbc_labeled.csv")

perform_eda(df, dataset_name="WDBC")
```

```python
# Train test split

from sklearn.model_selection import train_test_split

x = df.drop(columns=['diagnosis'])
y = df['diagnosis']

x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.2, random_state=42)
```

## Decision Tree Classifier

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import GridSearchCV, StratifiedKFold, cross_val_score
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score,
                              confusion_matrix, roc_curve, roc_auc_score, ConfusionMatrixDisplay)

# Baseline Decision Tree
dt_clf = DecisionTreeClassifier(random_state=42)
dt_clf.fit(x_train, y_train)

y_pred_dt = dt_clf.predict(x_test)
print("Baseline Decision Tree Accuracy:", accuracy_score(y_test, y_pred_dt))
```

### Hyperparameter Search Space - Decision Tree

```python
dt_param_grid = {
    'criterion': ['gini', 'entropy'],
    'max_depth': [3, 5, 7, 10, None],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4]
}

cv_strategy = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

dt_grid_search = GridSearchCV(
    estimator=DecisionTreeClassifier(random_state=42),
    param_grid=dt_param_grid,
    cv=cv_strategy,
    scoring='accuracy',
    n_jobs=-1
)

dt_grid_search.fit(x_train, y_train)

print("Best Decision Tree Parameters:", dt_grid_search.best_params_)
print("Best CV Accuracy:", dt_grid_search.best_score_)
```

### 5-Fold Cross-Validation Results Table - Decision Tree

```python
dt_cv_results = pd.DataFrame(dt_grid_search.cv_results_)

dt_summary = dt_cv_results[['param_criterion', 'param_max_depth', 'mean_test_score', 'std_test_score']]
dt_summary = dt_summary.rename(columns={
    'param_criterion': 'Criterion',
    'param_max_depth': 'Max Depth',
    'mean_test_score': 'Avg CV Accuracy',
    'std_test_score': 'Std CV Accuracy'
})

dt_summary = dt_summary.sort_values(by='Avg CV Accuracy', ascending=False).reset_index(drop=True)
dt_summary.head(10)
```

```python
# F1 scores for the same hyperparameter combinations
dt_f1_scores = []
for params in dt_cv_results['params']:
    model = DecisionTreeClassifier(random_state=42, **params)
    f1_scores = cross_val_score(model, x_train, y_train, cv=cv_strategy, scoring='f1_weighted')
    dt_f1_scores.append(f1_scores.mean())

dt_cv_results['mean_test_f1'] = dt_f1_scores
dt_full_summary = dt_cv_results[['param_criterion', 'param_max_depth', 'mean_test_score', 'mean_test_f1']]
dt_full_summary = dt_full_summary.rename(columns={
    'param_criterion': 'Criterion',
    'param_max_depth': 'Max Depth',
    'mean_test_score': 'Avg CV Accuracy (%)',
    'mean_test_f1': 'Avg CV F1 Score'
})
dt_full_summary['Avg CV Accuracy (%)'] = dt_full_summary['Avg CV Accuracy (%)'] * 100
dt_full_summary.sort_values(by='Avg CV Accuracy (%)', ascending=False).reset_index(drop=True)
```

### Train Final Decision Tree with Best Hyperparameters

```python
best_dt = dt_grid_search.best_estimator_
best_dt.fit(x_train, y_train)

y_pred_dt_best = best_dt.predict(x_test)

print("Tuned Decision Tree Accuracy:", accuracy_score(y_test, y_pred_dt_best))
print("Tuned Decision Tree F1 Score:", f1_score(y_test, y_pred_dt_best, average='weighted'))
```

## Random Forest Classifier

```python
from sklearn.ensemble import RandomForestClassifier

# Baseline Random Forest
rf_clf = RandomForestClassifier(random_state=42)
rf_clf.fit(x_train, y_train)

y_pred_rf = rf_clf.predict(x_test)
print("Baseline Random Forest Accuracy:", accuracy_score(y_test, y_pred_rf))
```

### Hyperparameter Search Space - Random Forest

```python
rf_param_grid = {
    'n_estimators': [50, 100, 200],
    'max_depth': [5, 10, None],
    'max_features': ['sqrt', 'log2'],
    'bootstrap': [True, False]
}

rf_grid_search = GridSearchCV(
    estimator=RandomForestClassifier(random_state=42),
    param_grid=rf_param_grid,
    cv=cv_strategy,
    scoring='accuracy',
    n_jobs=-1
)

rf_grid_search.fit(x_train, y_train)

print("Best Random Forest Parameters:", rf_grid_search.best_params_)
print("Best CV Accuracy:", rf_grid_search.best_score_)
```

### 5-Fold Cross-Validation Results Table - Random Forest

```python
rf_cv_results = pd.DataFrame(rf_grid_search.cv_results_)

rf_summary = rf_cv_results[['param_n_estimators', 'param_max_depth', 'param_max_features',
                             'param_bootstrap', 'mean_test_score', 'std_test_score']]
rf_summary = rf_summary.rename(columns={
    'param_n_estimators': 'n_estimators',
    'param_max_depth': 'Max Depth',
    'param_max_features': 'Max Features',
    'param_bootstrap': 'Bootstrap',
    'mean_test_score': 'Avg CV Accuracy',
    'std_test_score': 'Std CV Accuracy'
})

rf_summary = rf_summary.sort_values(by='Avg CV Accuracy', ascending=False).reset_index(drop=True)
rf_summary.head(10)
```

### Train Final Random Forest with Best Hyperparameters

```python
best_rf = rf_grid_search.best_estimator_
best_rf.fit(x_train, y_train)

y_pred_rf_best = best_rf.predict(x_test)

print("Tuned Random Forest Accuracy:", accuracy_score(y_test, y_pred_rf_best))
print("Tuned Random Forest F1 Score:", f1_score(y_test, y_pred_rf_best, average='weighted'))
```

## 5-Fold Cross-Validation Performance Comparison

```python
dt_fold_scores = cross_val_score(best_dt, x_train, y_train, cv=cv_strategy, scoring='accuracy')
rf_fold_scores = cross_val_score(best_rf, x_train, y_train, cv=cv_strategy, scoring='accuracy')

comparison_df = pd.DataFrame({
    'Model': ['Decision Tree', 'Random Forest'],
    'Fold 1': [dt_fold_scores[0], rf_fold_scores[0]],
    'Fold 2': [dt_fold_scores[1], rf_fold_scores[1]],
    'Fold 3': [dt_fold_scores[2], rf_fold_scores[2]],
    'Fold 4': [dt_fold_scores[3], rf_fold_scores[3]],
    'Fold 5': [dt_fold_scores[4], rf_fold_scores[4]],
})
comparison_df['Average'] = comparison_df[['Fold 1','Fold 2','Fold 3','Fold 4','Fold 5']].mean(axis=1)
comparison_df
```

## Evaluation Metrics

```python
def evaluate_model(y_true, y_pred, model_name):
    print(f"--- {model_name} ---")
    print("Accuracy :", accuracy_score(y_true, y_pred))
    print("Precision:", precision_score(y_true, y_pred, average='weighted'))
    print("Recall   :", recall_score(y_true, y_pred, average='weighted'))
    print("F1 Score :", f1_score(y_true, y_pred, average='weighted'))
    print()

evaluate_model(y_test, y_pred_dt_best, "Decision Tree")
evaluate_model(y_test, y_pred_rf_best, "Random Forest")
```

### Confusion Matrices

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

ConfusionMatrixDisplay.from_predictions(y_test, y_pred_dt_best, ax=axes[0], cmap='Blues')
axes[0].set_title("Decision Tree - Confusion Matrix")

ConfusionMatrixDisplay.from_predictions(y_test, y_pred_rf_best, ax=axes[1], cmap='Greens')
axes[1].set_title("Random Forest - Confusion Matrix")

plt.tight_layout()
save_fig("confusion_matrices.eps", fig=fig)
plt.show()
```

### ROC Curve and AUC

```python
y_test_bin = y_test.map({'B': 0, 'M': 1})
y_test_bin = y_test_bin.astype(int)

dt_probs = best_dt.predict_proba(x_test)[:, 1]
rf_probs = best_rf.predict_proba(x_test)[:, 1]

fpr_dt, tpr_dt, _ = roc_curve(y_test_bin, dt_probs)
fpr_rf, tpr_rf, _ = roc_curve(y_test_bin, rf_probs)

auc_dt = roc_auc_score(y_test_bin, dt_probs)
auc_rf = roc_auc_score(y_test_bin, rf_probs)

plt.figure(figsize=(7, 6))
plt.plot(fpr_dt, tpr_dt, label=f"Decision Tree (AUC = {auc_dt:.3f})")
plt.plot(fpr_rf, tpr_rf, label=f"Random Forest (AUC = {auc_rf:.3f})")
plt.plot([0, 1], [0, 1], linestyle='--', color='gray', label='Random Guess')
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve - Decision Tree vs Random Forest")
plt.legend()
save_fig("roc_curve.eps")
plt.show()
```

### Feature Importance - Random Forest

```python
importances = pd.Series(best_rf.feature_importances_, index=x.columns)
importances = importances.sort_values(ascending=False)

plt.figure(figsize=(10, 8))
sns.barplot(x=importances.values[:15], y=importances.index[:15])
plt.title("Top 15 Feature Importances - Random Forest")
plt.xlabel("Importance")
plt.tight_layout()
save_fig("feature_importance.eps")
plt.show()
```

## Observation Questions

**1. How does tree depth affect overfitting in Decision Trees?**

As `max_depth` increases, the Decision Tree fits the training data more closely, which increases training accuracy but tends to reduce test accuracy once the tree grows deep enough to memorize noise in the training set. Shallow trees underfit, while very deep (or unrestricted) trees overfit, showing a large gap between training and cross-validation accuracy.

**2. Which hyperparameter had the greatest impact on performance?**

For the Decision Tree, `max_depth` had the most noticeable effect on the bias-variance tradeoff. For the Random Forest, `n_estimators` and `max_features` had the greatest impact — increasing the number of trees generally stabilized accuracy, while `max_features` controlled the diversity among trees.

**3. How does Random Forest improve generalization?**

Random Forest builds many decision trees on bootstrapped samples of the data and randomly selects a subset of features at each split. Aggregating predictions from these diverse trees (via majority voting) reduces the variance of the overall model compared to a single Decision Tree, leading to better generalization on unseen data.

**4. Did ensemble learning always improve performance? Why or why not?**

In most cases Random Forest matched or outperformed the single Decision Tree, particularly in terms of stability across folds (lower variance in accuracy). However, the improvement in raw accuracy was sometimes marginal, since the Wisconsin Breast Cancer dataset is relatively clean and well-separated, so even a well-tuned single tree already performs well. Ensemble learning mainly helped reduce variance and overfitting rather than dramatically boosting raw accuracy.
